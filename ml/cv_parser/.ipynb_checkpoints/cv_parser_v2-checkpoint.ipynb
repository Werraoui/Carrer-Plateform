# CV Skill Parser — High-Performance Rebuild
## What changed and why
| Problem in original | Fix applied |
|---|---|
| `spacy.blank("en")` — no language knowledge | `en_core_web_lg` — pre-trained word vectors |
| `alignment_mode="contract"` — silently dropped 19% of training spans | `"expand"` — recovers those spans |
| Noisy labels (`"com"`, `"marital status"`, `"is"` as SKILL) | Filtering step removes structural/stop-word noise |
| `patience=1600`, no epoch cap — wasted 5 hrs | `patience=5`, `max_epochs=30`, early stopping |
| `width=96`, shallow encoder | `width=256`, `depth=4` — richer representations |
| No evaluation step | Full eval + error analysis cells included |
| PDF pages joined with `" "` — corrupts structure | `"\n".join(...)` — preserves layout |

## 2. Noise filter
The dataset has ~107 annotations per CV. Many are structural words (`"skills"`, `"work"`, `"com"`, 
`"marital status"`, `"is"`, `"c"`) that appear as SKILL labels but confuse the model.
This filter removes them so the model learns from clean signal only.

```python
# Words/phrases that appear as SKILL labels in the data but are not real skills.
# Identified by analysing the 30 most common single-token labels.
NOISE_SKILLS = {
    # structural CV words
    "skills", "work", "knowledge", "personal", "professional", "technical",
    "management", "company", "information", "training", "education",
    "qualification", "project", "system", "team", "office", "organization",
    "experience", "summary", "objective", "references", "declaration",
    "activities", "projects", "about", "profile", "contact",
    # noise tokens
    "com", "is", "ltd", "m", "c", "ms", "ph", "uc", "xp", "it", "rto",
    "email", "nationality", "gender", "passport",
    "marital status", "curriculum vitae", "work experience",
    "professional experience", "technical skills", "computer skills",
    "communication skills", "interpersonal skills", "professional qualification",
    "academic credentials",
    # generic soft skills (too vague — model can't learn boundaries)
    "good communication", "positive attitude", "quick learner", "team player",
    "team members", "team work", "hard work",
}

def is_noise(text: str) -> bool:
    t = text.strip().lower()
    if t in NOISE_SKILLS:
        return True
    if len(t) <= 1:          # single char
        return True
    if t.isdigit():          # pure number
        return True
    return False

print("Noise filter ready.")
print(f"Blocking {len(NOISE_SKILLS)} noisy patterns.")
```

## 3. Load and clean annotations

```python
import os, json

def load_spacy_data(folder_path):
    training_data = []
    skipped_noise = 0
    kept = 0

    for file_name in os.listdir(folder_path):
        if not file_name.endswith(".json"):
            continue
        with open(os.path.join(folder_path, file_name), "r", encoding="utf-8") as f:
            data = json.load(f)

        text = data["text"]
        entities = []
        for start, end, label in data["annotations"]:
            if "SKILL" not in label:
                continue
            span_text = text[start:end]
            if is_noise(span_text):
                skipped_noise += 1
                continue
            entities.append((start, end, "SKILL"))
            kept += 1

        if entities:
            training_data.append((text, {"entities": entities}))

    print(f"Documents loaded  : {len(training_data)}")
    print(f"Spans kept        : {kept}")
    print(f"Spans filtered    : {skipped_noise}  (noise labels)")
    return training_data

FOLDER = r"C:\Users\wiame\Desktop\career-platform\ml\cv_parser\ResumesJsonAnnotated"
data = load_spacy_data(FOLDER)
```

## 4. Quick data health check

```python
import spacy
from collections import Counter

nlp_blank = spacy.blank("en")
issues = Counter()
total = 0

for text, annots in data:
    doc = nlp_blank.make_doc(text)
    for start, end, _ in annots["entities"]:
        total += 1
        span = doc.char_span(start, end, alignment_mode="expand")
        if span is None:
            issues["null"] += 1
        elif span.text != span.text.strip():
            issues["whitespace"] += 1

print(f"Total spans : {total}")
print(f"Issues      : {dict(issues)}")
print(f"Drop rate   : {sum(issues.values())/total:.1%}  (target < 3%)")
```

## 5. Convert to spaCy format

```python
from spacy.tokens import DocBin
from tqdm import tqdm

def clean_text(text):
    if not isinstance(text, str):
        text = str(text)
    return text.encode("utf-8", errors="ignore").decode("utf-8", errors="ignore").replace("\x00", "")

def build_docbin(data, nlp, log_path):
    db = DocBin()
    stats = Counter()

    with open(log_path, "w", encoding="utf-8") as log:
        for text, annot in tqdm(data):
            text = clean_text(text)
            doc = nlp.make_doc(text)
            ents = []
            seen = set()

            for start, end, label in annot["entities"]:
                # trim whitespace at char level
                while start < end and text[start].isspace(): start += 1
                while end > start and text[end-1].isspace(): end -= 1
                if start >= end:
                    stats["empty"] += 1; continue

                span_idx = set(range(start, end))
                if span_idx & seen:
                    stats["overlap"] += 1; continue
                seen.update(span_idx)

                # FIX: "expand" not "contract" — recovers ~19% of dropped spans
                span = doc.char_span(start, end, label=label, alignment_mode="expand")
                if span is None:
                    log.write(f"[NULL] ({start},{end}) {repr(text[max(0,start-15):end+15])}\n")
                    stats["null"] += 1; continue

                ents.append(span)
                stats["ok"] += 1

            try:
                doc.ents = ents
                db.add(doc)
            except Exception as e:
                log.write(f"[DOC ERROR] {e}\n")
                stats["doc_error"] += 1

    print("Span stats:", dict(stats))
    return db
```

## 6. Train / test split and save

```python
from sklearn.model_selection import train_test_split
from pathlib import Path

# 85/15 — more training data → better generalisation for NER
train_data, test_data = train_test_split(data, test_size=0.15, random_state=42)
print(f"Train: {len(train_data)}  |  Test: {len(test_data)}")

BASE = Path(r"C:\Users\wiame\Desktop\career-platform\ml\cv_parser\model")
BASE.mkdir(parents=True, exist_ok=True)

# Use en_core_web_lg tokeniser for consistent tokenisation with the trained model
nlp_lg = spacy.load("en_core_web_lg", exclude=["ner", "tagger", "parser", "senter", "lemmatizer"])

print("\nConverting train...")
train_db = build_docbin(train_data, nlp_lg, BASE / "train_log.txt")
train_db.to_disk(BASE / "train_data.spacy")
print(f"✔ Saved train_data.spacy")

print("\nConverting test...")
test_db = build_docbin(test_data, nlp_lg, BASE / "test_log.txt")
test_db.to_disk(BASE / "test_data.spacy")
print(f"✔ Saved test_data.spacy")
```

## 7. Training config
**Key improvements over original:**
- `en_core_web_lg` vectors as base — the model inherits word embeddings
- `width=256` — richer token representations
- `patience=5` — stops early if no improvement (original was 1600 — never stopped early)
- `dropout=0.2` — reduces overfitting on repeated patterns
- `accumulate_gradient=4` — effective larger batch without more VRAM

## 10. Error analysis — understand what the model misses

```python
import spacy
from collections import Counter

nlp = spacy.load(r"C:\Users\wiame\Desktop\career-platform\ml\cv_parser\model\output\model-best")

false_negatives = []
false_positives = []
correct = []

for text, annots in test_data[:100]:
    gold = set(text[s:e].strip().lower() for s, e, _ in annots["entities"])
    pred_doc = nlp(text)
    pred = set(ent.text.strip().lower() for ent in pred_doc.ents)

    for m in gold - pred: false_negatives.append(m)
    for e in pred - gold: false_positives.append(e)
    for c in gold & pred: correct.append(c)

print(f"Correct predictions : {len(correct)}")
print(f"Missed (FN)         : {len(false_negatives)}")
print(f"Hallucinated (FP)   : {len(false_positives)}")
print()
print("Top 20 MISSED skills (annotate more of these):")
for skill, count in Counter(false_negatives).most_common(20):
    print(f"  {count:3d}×  {skill}")
print()
print("Top 20 HALLUCINATED skills (over-predicted):")
for skill, count in Counter(false_positives).most_common(20):
    print(f"  {count:3d}×  {skill}")
```

## 11. PDF skill extraction — test on real CVs

```python
import fitz  # PyMuPDF

def extract_skills_from_pdf(pdf_path: str, nlp) -> list[str]:
    """Extract unique skills from a CV PDF. Ready to use in your web app."""
    pdf = fitz.open(pdf_path)
    # FIX: join with newline, not space — preserves CV structure
    text = "\n".join(page.get_text() for page in pdf)
    doc = nlp(text)
    skills = sorted(set(
        ent.text.strip()
        for ent in doc.ents
        if ent.label_ == "SKILL" and len(ent.text.strip()) > 1
    ))
    return skills

# Test on all CVs in your test folder
import os
TEST_DIR = r"C:\Users\wiame\Desktop\career-platform\ml\cv_parser\test"

for fname in os.listdir(TEST_DIR):
    if not fname.endswith(".pdf"):
        continue
    path = os.path.join(TEST_DIR, fname)
    skills = extract_skills_from_pdf(path, nlp)
    print(f"\n{'='*50}")
    print(f"  {fname}")
    print(f"  {len(skills)} skills found")
    print(f"{'='*50}")
    for s in skills:
        print(f"  • {s}")
```

## 12. Web app integration
Drop these two functions into your FastAPI or Flask backend.

```python
import spacy, fitz
from pathlib import Path

# ── Singleton loader — load once, reuse forever ──────────────────────────────
_nlp_cache = {}

def get_model(model_path: str):
    if model_path not in _nlp_cache:
        _nlp_cache[model_path] = spacy.load(model_path)
    return _nlp_cache[model_path]


# ── Main API functions ────────────────────────────────────────────────────────
def extract_skills_from_text(text: str, model_path: str) -> list[str]:
    """Return deduplicated, sorted skill list from raw text."""
    nlp = get_model(model_path)
    doc = nlp(text)
    return sorted(set(
        ent.text.strip()
        for ent in doc.ents
        if ent.label_ == "SKILL" and len(ent.text.strip()) > 1
    ))


def extract_skills_from_pdf(pdf_path: str, model_path: str) -> dict:
    """Open PDF, extract text, return skills + metadata."""
    pdf = fitz.open(pdf_path)
    pages_text = [page.get_text() for page in pdf]
    full_text = "\n".join(pages_text)
    skills = extract_skills_from_text(full_text, model_path)
    return {
        "skills": skills,
        "skill_count": len(skills),
        "pages": len(pages_text),
        "char_count": len(full_text),
    }


# ── Smoke test ────────────────────────────────────────────────────────────────
MODEL_PATH = r"C:\Users\wiame\Desktop\career-platform\ml\cv_parser\model\output\model-best"
PDF_PATH   = r"C:\Users\wiame\Desktop\career-platform\ml\cv_parser\test\CV_Erraoui_Wiame.pdf"

result = extract_skills_from_pdf(PDF_PATH, MODEL_PATH)
print(json.dumps(result, indent=2))
```
